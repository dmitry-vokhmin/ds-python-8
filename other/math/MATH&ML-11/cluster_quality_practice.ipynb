```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score, adjusted_rand_score, homogeneity_score, completeness_score
```

```python
X_test = pd.read_csv('data/test.txt', sep='\s+', header=None)
X_train = pd.read_csv('data/train.txt', sep='\s+', header=None)
y_train = pd.read_csv('data/train_labels.txt', header=None)
y_test = pd.read_csv('data/test_labels.txt', header=None)
```

```python
X = pd.concat([X_train, X_test])
y = pd.concat([y_train, y_test])
X.shape
```

```python
y.nunique()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[0][0]
```

```python
def get_metric_score(cluster_num, X, metric):
    k_means = KMeans(n_clusters=cluster_num, random_state=42)
    k_means.fit(X)
    metric_score = metric(X, k_means.predict(X))
    return metric_score
```

```python
metric_res = {}

for metric in [silhouette_score, calinski_harabasz_score, davies_bouldin_score]:
    if metric.__name__ not in metric_res:
        metric_res[metric.__name__] = {'metric': [], 'cluster': []}
    for cluster_num in range(2, 10):
        metric_res[metric.__name__]['metric'].append(get_metric_score(cluster_num, X_scaled, metric))
        metric_res[metric.__name__]["cluster"].append(cluster_num)
    metric_res[metric.__name__]['df'] = pd.DataFrame(metric_res[metric.__name__])
```

```python
k_means = KMeans(n_clusters=6, init='random', random_state=42)
k_means.fit(X_scaled)

y_pred = k_means.predict(X_scaled)

print('Hom:', round(homogeneity_score(y[0].values, y_pred), 2))
print('Com:', round(completeness_score(y[0].values, y_pred), 2))
print('Ran:', round(adjusted_rand_score(y[0].values, y_pred), 2))
```

```python
ct = pd.crosstab(y[0], y_pred)
```

```python
k_means = KMeans(n_clusters=2, init='random', random_state=42)
k_means.fit(X_scaled)

y_pred = k_means.predict(X_scaled)

print('Hom:', round(homogeneity_score(y[0].values, y_pred), 2))
print('Com:', round(completeness_score(y[0].values, y_pred), 2))
print('Ran:', round(adjusted_rand_score(y[0].values, y_pred), 2))
```
