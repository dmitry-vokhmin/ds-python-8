```python
import numpy as np


def f(x, y):
    return (1 - x) ** 2 + 100 * (y - x ** 2) ** 2
def grad(x, y):
    dx = 2 * x - 2 + 100 * (-4 * y * x + 4 * x ** 3)
    dy = 100 * (2 * y - 2 * x ** 2)
    return (dx, dy)
```

```python
x0 = (0, 0)
gamma = 0.001
x_cur = x0

vals = []
coords = []
i = 0
while True:
    x_new = (x_cur[0] - gamma * grad(*x_cur)[0],
            x_cur[1] - gamma * grad(*x_cur)[1])
    if i > 50000:
        break
    x_cur = x_new
    vals.append(f(*x_cur))
    coords.append(x_cur)
    i += 1
    print(f"iter={i}; x=({x_cur[0]:.4f}, {x_cur[1]:.4f});"
          f" f(x)={f(*x_cur):.4f}; grad f(x)=({grad(*x_cur)[0]:.4f}, {grad(*x_cur)[1]:.4f})")
```

```python
from scipy import optimize
```

```python
optimize.minimize(lambda x: f(*x), x0=(0, 0))
```

```python
def find_best_res(f, n_vals):
    best_res = None
    np.random.seed(42)

    # Попробуем несколько случайных стартов
    for start in np.random.uniform(-2, 2, size=(5, n_vals)):
        res = optimize.minimize(f, start)
        if best_res is None or res.fun < best_res.fun:
            best_res = res


    return best_res
```

```python
def f_uv(vars):
    x, y = vars
    return 2 * x**2 - 4 * x * y + y**4 + 2

find_best_res(f_uv, 2)['fun']
```

```python
def f_uv(vars):
    x, y, z = vars
    return x**3 - 2*x**2 + y**2 + z**2 - 2*x*y + x*z - y*z + 3*z

find_best_res(f_uv, 3)
```
