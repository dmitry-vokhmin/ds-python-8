```python
# сумма векторов и умножение на число
a=np.array([1,2,3])
b=np.array([6,7,8])
print('a+b:',a+b)
print('2a:',2*a)
```

```python
# сумма списков и умножение на число не является векторной операцией!
a_list=[1,2,3]
b_list=[6,7,8]
print('a+b:',a_list+b_list)
print('2a:',2*a_list)
```

```python
# линейная комбинация с коэффициентами 2, 2, 3
a=np.array([1,2,3])
b=np.array([6,7,8])
c=np.array([1,0,1])

d=2*a+2*b+3*c
print('linear combination:',d)
```

```python
# скалярное произведение

s=np.dot(a,b)
print('scalar product:',s)

# покоординатное умножение
print('coordinate product:',a*b)
```

```python
# скалярное произведение - еще один способ
s1=a@b
print('scalar product:',s1)
```

```python
# длина вектора
l=np.linalg.norm(a)
print('length of a:',l)
```

```python
# нормировка
a_norm=a/np.linalg.norm(a)
print('a_norm:',a_norm)
print('length of a_norm:',np.linalg.norm(a_norm))
```

```python
#пропорциональность и ортогональность
a=np.array([1,2,1])
b=a*2
c=np.array([-1,1,-1])
print('a:',a)
print('b:',b)
print('c:',c)
```

```python
#пропорциональность и ортогональность
print("(a,b):",np.dot(a,b))
print("(a,c):",np.dot(a,c))
print("(b,c):",np.dot(b,c))
```

```python
# попарная ортогональность - да
a=np.array([1,0,1])
b=np.array([0,1,0])
c=np.array([1,0,-1])
print('a:',a)
print('b:',b)
print('c:',c)
```

```python
# попарная ортогональность - да
print('(a,b):',np.dot(a,b))
print('(b,c):',np.dot(b,c))
print('(a,c):',np.dot(a,c))
```

```python
# попарная ортогональность - нет
c1=np.array([1,0,4])
print('a:',a)
print('b:',b)
print('с1:',c1)
```

```python
# попарная ортогональность - нет
print('(a,b):',a@b)
print('(b,c1):',b@c1)
print('(a,c1):',a@c1)
```

```python
from sklearn.preprocessing import PolynomialFeatures

x = np.array([[1, 3, -2, 9]]).T
y = np.array([3, 7, -5, 21])
poly = PolynomialFeatures(degree=2, include_bias=True)
poly_x = poly.fit_transform(x)
w_hat = np.linalg.inv(poly_x.T@poly_x)@poly_x.T@y
w_hat
```

```python
from sklearn.linear_model import Ridge

x0 = np.ones(5)
x1 = [5, 9, 4, 3, 5]
x2 = [15, 18, 18, 19, 19]
x3 = [7, 6, 7, 7, 7]
y = np.array([24, 22, 35, 33, 36])

alpha = 1

X = np.array([x0, x1, x2, x3]).T

model = Ridge(alpha=alpha, fit_intercept=False)
model.fit(X, y)
print(model.coef_)
```
