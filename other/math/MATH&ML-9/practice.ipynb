```python
import pandas as pd
from sklearn.ensemble import BaggingClassifier, RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.metrics import f1_score, mean_absolute_error
```

```python
wine_data = pd.read_csv('wineQualityReds.csv', index_col=0)
wine_data.head()
```

```python
wine_data['quality'] = wine_data['quality'].apply(lambda x: 1 if x >= 6 else 0)
X = wine_data.drop('quality', axis=1)
y = wine_data['quality']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
lg = LogisticRegression(max_iter=1000)
lg.fit(X_train, y_train)

y_lg_pred = lg.predict(X_test)
```

```python
dt = DecisionTreeClassifier(random_state=42, max_depth=10)
dt.fit(X_train, y_train)

y_dt_pred = dt.predict(X_test)
```

```python
print("LogisticRegression F1-score\n", round(f1_score(y_test, y_lg_pred), 3))
print("DecisionTreeClassifier F1-score\n", round(f1_score(y_test, y_dt_pred), 3))
```

```python
dt = DecisionTreeClassifier(random_state=42, max_depth=10)
bc = BaggingClassifier(estimator=dt, n_estimators=1500, random_state=42)
bc.fit(X_train, y_train)

y_bc_pred = bc.predict(X_test)
print("BaggingClassifier F1-score\n", round(f1_score(y_test, y_bc_pred), 3))
```

## Случайный лес

```python
house_data = pd.read_csv('boston.csv')
house_data.head()
```

```python
for col in house_data.columns:
    if house_data[col].dtype == 'object':
        house_data[col] = pd.to_numeric(house_data[col].str.replace(',', '.', regex=False))
X = house_data.drop(columns=['target'])
y = house_data['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=13)
```

```python
print("Среднее значение медианных цен:", round(y_train.mean(), 2))
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

y_pred = lin_reg.predict(X_test)

print('MAE:', round(mean_absolute_error(y_test, y_pred), 2))
```

```python
dt = DecisionTreeRegressor(random_state=13)
dt.fit(X_train, y_train)

y_train_pred = dt.predict(X_train)
y_test_pred = dt.predict(X_test)

print('MAE train:', round(mean_absolute_error(y_train, y_train_pred), 2))
print('MAE test:', round(mean_absolute_error(y_test, y_test_pred), 2))
```

```python
rf_3 = RandomForestRegressor(n_estimators=3, random_state=13)
rf_10 = RandomForestRegressor(n_estimators=10, random_state=13)
rf_100 = RandomForestRegressor(n_estimators=100, random_state=13)
rf_500 = RandomForestRegressor(n_estimators=500, random_state=13)

rf_3.fit(X_train, y_train)
rf_10.fit(X_train, y_train)
rf_100.fit(X_train, y_train)
rf_500.fit(X_train, y_train)

y_3_pred = rf_3.predict(X_test)
y_10_pred = rf_10.predict(X_test)
y_100_pred = rf_100.predict(X_test)
y_500_pred = rf_500.predict(X_test)

print('MAE 3:', round(mean_absolute_error(y_test, y_3_pred), 2))
print('MAE 10:', round(mean_absolute_error(y_test, y_10_pred), 2))
print('MAE 100:', round(mean_absolute_error(y_test, y_100_pred), 2))
print('MAE 500:', round(mean_absolute_error(y_test, y_500_pred), 2))
```
